````python
# function to generate, process, and return the response from the LLM
def generate_llama_response1(user_prompt):

    # System message
    system_message = """
    [INST]<<SYS>> Respond to the user question based on the user prompt<</SYS>>[/INST]
    """

    # Combine user_prompt and system_message to create the prompt
    prompt = f"{user_prompt}\n{system_message}"

    # Generate a response from the LLaMA model
    response = lcpp_llm(
        prompt=prompt,
        max_tokens=1024,
        temperature=0.01,
        top_p=0.95,
        repeat_penalty=1.2,
        top_k=50,
        stop=['INST'],
        echo=False
    )

    # Extract and return the response text
    response_text = response["choices"][0]["text"]
    return response_text
````

````python
def generate_llama_response(user_prompt, max_tokens=256, temperature=0.1):
    """
    Generate response from BioMistral-7B without RAG.

    Args:
        user_prompt: User's medical question
        max_tokens: Maximum tokens to generate
        temperature: Sampling temperature (0 = deterministic)

    Returns:
        Generated response text
    """
    # Correct Mistral format: question INSIDE [INST] tags
    prompt = f"""[INST] You are a knowledgeable medical assistant. Answer the following question accurately and concisely based on your medical knowledge.

{user_prompt} [/INST]"""

    try:
        response = lcpp_llm(  # or lcpp_llm if that's your variable name
            prompt=prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=0.95,
            top_k=50,
            repeat_penalty=1.1,
            stop=['[INST]', '</s>'],
            echo=False
        )

        return response["choices"][0]["text"].strip()

    except Exception as e:
        return f"Error: {e}"
````

## Question Answering using LLM with Prompt Engineering

````python
def generate_llama_response_engineered_sections(user_prompt):
    """
    Generate response section by section to avoid context window limits.
    Each section gets its own generation with full context window available.
    """
    sections = [
        "CLINICAL PRESENTATION",
        "DIAGNOSIS",
        "TREATMENT",
        "SPECIAL CONSIDERATIONS"
    ]

    full_response = ""

    for section in sections:
        section_prompt = f"""[INST] You are an expert clinical medical doctor.

Question: {user_prompt}

Provide ONLY the {section} section with comprehensive detail. Include all relevant information for this section.

{section}: [/INST]"""

        response = lcpp_llm(
            prompt=section_prompt,
            max_tokens=2048,  # Plenty of room for each section
            temperature=0.25,
            top_p=0.95,
            top_k=50,
            repeat_penalty=1.05,
            stop=['[INST]', '</s>', '\n\n2.', '\n\n3.', '\n\n4.'],  # Stop if it tries to continue to next section
            echo=False
        )

        section_text = response['choices'][0]['text'].strip()
        full_response += f"{section}:\n{section_text}\n\n"

        print(f"✓ Generated {section} ({len(section_text)} chars)")

    return full_response


````

````python
def generate_llama_response_conditional_sections(user_prompt, patient_age=None, is_emergency=False):
    """
    LESSON 2: Conditional prompting with section-by-section generation
    Fixes the context window limitation by generating each section separately
    """

    # Build conditional instructions
    urgency_instruction = ""
    if is_emergency:
        urgency_instruction = "URGENT CASE: Prioritize immediate life-saving interventions and time-sensitive treatments."

    age_instruction = ""
    if patient_age:
        if patient_age < 18:
            age_instruction = "PEDIATRIC PATIENT: Adjust all dosages, diagnostic approaches, and treatment for pediatric population."
        elif patient_age > 65:
            age_instruction = "ELDERLY PATIENT: Consider polypharmacy, comorbidities, and age-related changes."

    sections = [
        "CLINICAL PRESENTATION",
        "DIAGNOSIS",
        "TREATMENT",
        "SPECIAL CONSIDERATIONS"
    ]

    full_response = ""

    # Add context header if conditions apply
    if urgency_instruction or age_instruction:
        full_response += f"{'[EMERGENCY] ' if is_emergency else ''}"
        full_response += f"{'[Age: ' + str(patient_age) + '] ' if patient_age else ''}\n\n"

    for section in sections:
        section_prompt = f"""[INST] You are an expert clinical medical doctor.

{urgency_instruction}
{age_instruction}

Question: {user_prompt}

Provide ONLY the {section} section with comprehensive detail.

{f"Remember: Patient age is {patient_age} years old." if patient_age else ""}
{f"Remember: This is an EMERGENCY - prioritize urgent interventions." if is_emergency else ""}

{section}: [/INST]"""

        response = lcpp_llm(
            prompt=section_prompt,
            max_tokens=2048,
            temperature=0.25,
            top_p=0.95,
            top_k=50,
            repeat_penalty=1.05,
            stop=['[INST]', '</s>', '\n\n2.', '\n\n3.', '\n\n4.'],
            echo=False
        )

        section_text = response['choices'][0]['text'].strip()
        full_response += f"{section}:\n{section_text}\n\n"

        print(f"✓ Generated {section} ({len(section_text)} chars)")

    return full_response

````

Why this is Lesson 2 (Conditional):

✅ AI behavior changes based on input parameters

✅ Different instructions for different scenarios

✅ Adapts response based on conditions (age, urgency)

````python
def generate_llama_response_stepwise(user_prompt, max_tokens=3000):
    """
    LESSON 2: Step-wise expectations - guiding AI through a reasoning process
    """

    prompt = f"""[INST] You are an expert clinical medical doctor. Answer the following question by thinking through it step-by-step:

Question: {user_prompt}

Follow these steps in your response:

STEP 1: IDENTIFY THE CONDITION
First, clearly define the medical condition and its pathophysiology in 2-3 sentences.

STEP 2: RECOGNIZE THE PRESENTATION
Then, describe how this condition typically presents clinically, including key signs and symptoms.

STEP 3: ESTABLISH THE DIAGNOSIS
Next, outline the diagnostic approach including physical examination findings, laboratory tests, and imaging studies needed.

STEP 4: DETERMINE THE TREATMENT
After that, detail the treatment protocol including immediate interventions, ongoing management, and expected outcomes.

STEP 5: CONSIDER SPECIAL CASES
Finally, address any special populations, complications, or important considerations that could affect management.

Begin your response now, following each step in order:

[/INST]"""

    response = lcpp_llm(
        prompt=prompt,
        max_tokens=max_tokens,
        temperature=0.25,
        top_p=0.95,
        top_k=50,
        repeat_penalty=1.05,
        stop=['[INST]', '</s>'],
        echo=False
    )

    return response["choices"][0]["text"].strip()
````

Why this is (Step-wise):

✅ Guides AI through logical reasoning sequence

✅ Forces structured thinking process

✅ Each step builds on previous step

````python
import json
import re

def generate_llama_response_json_sections(sections_dict):
    """
    Generate JSON response section by section for reliability.
    """
    result = {}

    for section_key, section_description in sections_dict.items():
        section_prompt = f"""[INST] You are an expert clinical medical doctor.

Generate content for the "{section_key}" section: {section_description}

Return ONLY a JSON object with key "{section_key}" containing relevant information.
DO NOT use markdown. Return pure JSON only.

[/INST]"""

        response = lcpp_llm(
            prompt=section_prompt,
            max_tokens=1500,
            temperature=0.25,
            top_p=0.95,
            top_k=50,
            repeat_penalty=1.05,
            stop=['[INST]', '</s>'],
            echo=False
        )

        response_text = response["choices"][0]["text"].strip()
        response_text = re.sub(r'```json\s*', '', response_text)
        response_text = re.sub(r'```\s*$', '', response_text)
        response_text = response_text.strip()

        try:
            section_json = json.loads(response_text)
            result.update(section_json)
            print(f"✓ Generated {section_key}")
        except json.JSONDecodeError:
            print(f"⚠️ {section_key} not valid JSON")
            result[section_key] = response_text

    return json.dumps(result, indent=2)

# Define sections
sections = {
    "clinical_presentation": "signs/symptoms, injury mechanisms, severity assessment, neurovascular evaluation",
    "diagnosis": "field assessment, clinical examination, imaging, classification systems",
    "treatment": "field care, transport, definitive treatment, rehabilitation, timeline",
    "special_considerations": "field challenges, complications, healing factors, return-to-activity"
}

fractured_leg_json = generate_llama_response_json_sections(sections)
print("\nFinal JSON:")
print(fractured_leg_json)
````

````python
def generate_rag_response(user_input,k=3,max_tokens=128,temperature=0,top_p=0.95,top_k=50):
    global qna_system_message,qna_user_message_template
    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input,k=k)
    context_list = [d.page_content for d in relevant_document_chunks]

    # Combine document chunks into a single context
    context_for_query = ". ".join(context_list)

    user_message = qna_user_message_template.replace('{context}', context_for_query)
    user_message = user_message.replace('{question}', user_input)

    prompt = qna_system_message + '\n' + user_message

    # Generate the response
    try:
        response = llm(
                  prompt=prompt,
                  max_tokens=max_tokens,
                  temperature=temperature,
                  top_p=top_p,
                  top_k=top_k
                  )

        # Extract and print the model's response
        response = response['choices'][0]['text'].strip()
    except Exception as e:
        response = f'Sorry, I encountered the following error: \n {e}'

    return response
````
